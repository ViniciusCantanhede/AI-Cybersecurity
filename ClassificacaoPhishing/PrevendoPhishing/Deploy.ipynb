## Deploy prevendo ataques de phishing

```python
import numpy as np 
import pandas as pd
from joblib import load
```

```python
# Vamos carregar os modelos do disco
modelo_pca_deploy = load('modelo/modelo_pca.joblib') 
modelo_classificador_deploy = load('modelo/modelo_versao_final.joblib') 
```

```python
# Carregamos o dataset com dados de uma nova url
dados = pd.read_csv('dados/novos_dados.csv', header = 0)
```

```python
dados.head()
```

```python
# Convertemos o dataframe em array numpy (pois foi com esse formato que treinamos o modelo PCA)
dados_array = dados.to_numpy()
```

```python
dados_array_pca = modelo_pca_deploy.transform(dados_array) 
previsao = modelo_classificador_deploy.predict(dados_array_pca)  
print('Para a URL apresentada o modelo previu a seguinte classe:', previsao)
```

Conclusão: A URL apresentada é legítima e não representa ataque de phishing.
